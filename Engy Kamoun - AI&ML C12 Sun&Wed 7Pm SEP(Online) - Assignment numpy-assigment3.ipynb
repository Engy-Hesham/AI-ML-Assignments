1. Create a 3x3 array with random integers between 1 and 100. Write a program to "normalize" the array so that the values are scaled between 0 and 1. 
- Formula: (x - min) / (max - min) 
- Example Output: An array where the smallest value becomes 0.0 and the largest becomes 1.0.

````python
arr_q1 = np.random.randint(1, 101, (3, 3))
min_val = np.min(arr_q1)
max_val = np.max(arr_q1)

arr_n_q1 = (arr_q1 - min_val) / (max_val - min_val)

arr_n_q1
````

2. Create a 1D array of 10 random numbers. Ask the user for a value X Find the value in the array that is closest to X 
- Example Input: Array: [1, 5, 8, 12], X = 10 
- Example Output: 8 or 12 (whichever is mathematically closer).

````python
arr_q2 = np.random.randint(1,20,10)
x = int(input("Enter a number")) #10 is the input value
arr_diff_q2 = np.abs(arr_q2-x)
print(f"Array: {arr_q2}\nInput:{x}\nClosest Value: {arr_q2[np.argmin(arr_diff_q2)]}")
````

3. Write a NumPy program to create a 5x5 array with random values. For each column in the array, find the second-largest value. 
- Example Input: [[1, 2, 3], [4, 5, 6], [7, 8, 9]] 
- Example Output: [4, 5, 6] (These are the second-highest values for each respective column).

````python
arr_q3 = np.random.randint(1,100,(5,5))
print(f"Original Array: \n {arr_q3}")
print(f"Second highest values in each column: {np.sort(arr_q3, axis=0)[-2,:]}")
````

4. Create a 6x4 matrix of random integers. Identify which rows have a sum that is an even number. Print the indices of these rows.

````python
arr_q4 = np.random.randint(1, 100, (6,4))
np.sum(arr_q4, axis=1)
print(f"Array: \n{arr_q4}")
arr_sum_q4 = np.sum(arr_q4, axis=1)
print(f"Summed Rows Array: \n{arr_sum_q4}")
index_vals = np.where(arr_sum_q4 % 2 == 0)[0]
print(f"Rows index: {index_vals}")
````

5. Create a 1D array of 30 random integers between 1 and 50. Replace all numbers that are multiples of 3 or 5 with the value -1. 
- Example Input: [3, 7, 10, 15] 
- Example Output: [-1, 7, -1, -1]

````python
arr_q5 = np.random.randint(1, 51, 30)
condition_q5 = (arr_q5 % 3 == 0) | (arr_q5 % 5 == 0)
print(f"Array:\n{arr_q5}\n\nMultiples of 3 or 5: \n{arr_q5[np.where(condition_q5)]}\n")
arr_q5[condition_q5] = -1
print(f"Array after values change:\n{arr_q5}")
````

6. Write a NumPy program to create a 3x3 array with random values. For each row, calculate its mean (average) and subtract that mean from every element in that specific row. 
- Example Input: ``` [[10, 20, 30], [5, 5, 5], [0, 10, 20]] 
- Example Output: [[-10. 0. 10.], [ 0. 0. 0.], [-10. 0. 10.]]

````python
arr_q6 = np.random.randint(1, 100, (3,3))
arr_m_q6 = np.mean(arr_q6, axis=1, keepdims=True) #.reshape(3,1) could have been used instead of keepdims=True
print(f"Array:\n{arr_q6}\n\nMean Array:\n{arr_m_q6}\n\nArray after subtraction of mean:\n{arr_q6 - arr_m_q6}")
````
