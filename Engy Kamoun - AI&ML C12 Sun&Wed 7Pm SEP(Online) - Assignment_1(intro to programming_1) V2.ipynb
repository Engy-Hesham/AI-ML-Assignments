1. Ask the user for a string that contains both letters and numbers (e.g., "Py2th0on3"). Write a program that separates them into two lists: one for the letters and one for the numbers. Calculate the sum of the numbers found and join the letters to form a word. 
- Example Input: "Py2th0on3" 
- Example Output: Word: "Python", Sum: 5

```python
text = input("Enter text: ")
letters, numbers = [], []
for char in text:
    if char.isalpha():
        letters.append(char)
    elif char.isdigit():
        numbers.append(int(char))

joined_letters = ''.join(letters)
print(f'Word: "{joined_letters}", Sum: {sum(numbers)}')
```

2. Ask a user to enter a password. Check if it is valid based on these rules: Length must be between 8 and 16 characters, and it must contain at least one uppercase letter and one special character (e.g., @, #, $). 
- Example Input: "Guest@123" 
- Example Output: "Password is Secure."

```python
password = input("Enter password \nLength between 8 and 16 characters \nOne uppercase letter \nOne special character: ") 
#Input doesn't produce new line between text as in code for some reason

is_len_valid = len(password) >= 8 and len(password) <= 16
#This could have been done in one line but I think this aligns more with what we have taken till now
has_upper, has_special = False, False

#Iterate through each character in the password to check for uppercase letters and special characters
#Loops until both conditions are met or until all characters have been checked
for char in password:
    if not has_upper and char.isupper():
        has_upper = True

    if not has_special and not char.isalnum():
        has_special = True

    if has_upper and has_special:
        break


if is_len_valid and has_upper and has_special:
    print("Password is Secure.")
else:
    print("Please enter a valid password.")
```

3. A store gives discounts based on the total price: Under $50 (No discount), $50 to $100 (10% discount), and Over $100 (20% discount). If the user is a "Member," they get an additional 5% off the final calculated price. 
- Example Input: Price: 100, Member: Yes 
- Example Output: Final Total: $85.5 (Note: 100 - 10% = 90; then 90 - 5% = 85.5)

```python
user_input = input("Enter price and if the customer is a member (eg: Price: 100, Member: Yes): ")

# Decided to go with this style here as it could have been a proper for loop to strip the spaces in each element in the list
input_list = [val.strip() for val in user_input.replace(",", ":").split(":")]
price, member = int(input_list[1]), input_list[3]

if price < 50:
    total_price = price
elif price >= 50 and price <= 100:
    total_price = price * 0.9  # For the 10% discount
else:
    total_price = price * 0.8  # For the 20% discount

if member.lower() == "yes":
    total_price *= 0.95  # For the additional 5% discount for members

print(f"Final Total: ${total_price}")
```

4. Ask the user to input two separate lists of names (comma-separated). Create a third list that contains only the names that appear in both lists. 
- Example Input: List 1: "Ali, Ben, Sara", List 2: "Ben, Joy, Ali" 
- Example Output: Common names: ['Ali', 'Ben']

```python
list1 = input("Enter first list of names separated by commas: ").lower().split(",")
list2 = input("Enter second list of names separated by commas: ").lower().split(",")

common_names = []

list1 = [name.strip() for name in list1]
list2 = [name.strip() for name in list2]

for name in list1:
    if name in list2:
        common_names.append(name)

print("Common names:", [name.capitalize() for name in common_names])
```

5. Create a dictionary where keys are student names and values are lists of 3 test scores. Write a program that calculates the average for each student. If the average is > 90, print the student's name followed by "Honor Roll."

- Example Output: Sam: Average 92.3 - Honor Roll

```python
students_dict = {'Sam':[95, 90, 92], 'John':[88, 85, 91], 'Alice':[92, 89, 95], 'Bob':[78, 82, 80]}

# Calculate the average grade for each student
for student, grades in students_dict.items():
    average = sum(grades) / len(grades)
    if average > 90: #After mentor check, '=' sign was removed
        print(f"{student}: Average {average:.1f} - Honor Roll")
```

- Example Input: word1 = "abc", word2 = "xyz"

8. Given a list of numbers with many duplicates, return a list of tuples showing each unique element and its frequency (how many times it appeared). 
- Example Input: [1, 1, 2, 3, 3, 3, 4] 
- Example Output: [(1, 2), (2, 1), (3, 3), (4, 1)]

```python
dup_list = input("Enter a list of numbers separated by commas: ").replace("[", "").replace("]", "").split(",")

# Using list comprehension and set to get unique values and their counts
dup_list = [int(x.strip()) for x in dup_list]
dedup_list = [(x, dup_list.count(x)) for x in set(dup_list)]
print(dedup_list)

```

9. Ask the user for a number $n$. Calculate the factorial (n!) using a loop, and then check if the resulting factorial is divisible by the sum of numbers from 1 to $n$. 
- Example Input: 4 
- Example Output: Factorial is 24, Sum is 10. 24 is not divisible by 10.

```python
number_input = int(input("Enter a number: "))
factorial, summed = 1, 0
divisible = 'is not divisible'

for num in range(number_input,0,-1):
    factorial *= num
    summed += num

if factorial % summed == 0:
    divisible = 'is divisible'

print(f"Factorial is {factorial}, Sum is {summed}. {factorial} {divisible} by {summed}")
```

**New Solution**

```python
store_dict = {'apple': 0.5, 'bread': 2.0, 'milk': 1.5}

# A dictionary that holds possible mentioned actions with their meaning
action_synonyms = {
    "add": "add",
    "buy": "add",
    "purchase": "add",
    "insert": "add",
    "remove": "remove",
    "delete": "remove",
    "take out": "remove",
    "discard": "remove"
}

entries = []
more_input = True

while more_input:
    entries.append(input("Add to or remove items from cart (eg: Add 'bread', Remove 'milk')"))
    more_input = input('Anything else? y/n').lower() == 'y'

final_cart = {}

def parse_command(prompt):
    #To remove additional spaces between words, make them lowercase and then rejoin them
    normalized = " ".join(prompt.lower().split())

    commands = [command.strip() 
                for command in normalized.replace(" and ", ",").split(",")] # in case 'and' is used instead of ','

    parsed_commands = []
    current_action = None

    for command in commands:
        # Remove please word if mentioned in command
        command = command.removeprefix("please ").removesuffix(" please").strip()

        # Reject empty part of command
        if not command:
            return None

        explicit_action = None
        item_text = command

        # Find approved action word
        for phrase, meaning in action_synonyms.items():
            if command == phrase:
                explicit_action = phrase
                item_text = ""
                break
            elif command.startswith(phrase + " "):
                explicit_action = phrase
                item_text = command[len(phrase):].strip()
                break

        # Use the found action or take the action before it (for case like: buy milk and bread)
        if explicit_action is not None:
            current_action = action_synonyms[explicit_action]
        elif current_action is None:
            return None

        # check item existance after an action
        if not item_text:
            return None

        # If there are quotes around item, then make sure they match (single or double quotes)
        if item_text[0] in ("'", '"') or item_text[-1] in ("'", '"'):
            quote = item_text[0]
            if len(item_text) < 2 or item_text[-1] != quote:
                return None

            item = item_text[1:-1].strip()  

            # This is an assumption that item text doesn't allow quote marks inside of it (Not the surronding quotes)
            if "'" in item or '"' in item:
                return None

        else:
            # This means the item name is unquoted so another assumption is that items must be one word and have no quote marks
            if "'" in item_text or '"' in item_text or len(item_text.split()) != 1:
                return None
            item = item_text

        # To reject empty items
        if not item:
            return None    

        parsed_commands.append((current_action, item))

    return parsed_commands

for entry in entries:
    parsed_commands = parse_command(entry)

    if parsed_commands is None:
        print(f"Invalid input: {entry}. Check the action, item and separators.")
        continue

    for action, item in parsed_commands:
        if action == 'add':
            if item not in store_dict:
                print(f"{item} isn't available")
            else:
                final_cart[item] = store_dict[item]
        elif item in final_cart:
            final_cart.pop(item)
        else:
            print(f"{item} isn't in your cart")

total_bill = sum(final_cart.values())
print(f"Total Bill: ${total_bill}")
```

11. Check if a full sentence is a palindrome, ignoring spaces and capitalization. 
- Example Input: "Race car" 
- Example Output: True

```python
sentence = input('Enter a sentence: ').replace(' ', '').lower()
rev_sentence = ''

for i in range(len(sentence)-1,-1,-1):
    rev_sentence += sentence[i]

print(sentence == rev_sentence)
```

14. Take a dictionary like {'A': 1, 'B': 2, 'C': 1} and flip it. If two keys have the same value, the new value should be a list of those keys. 
- Example Input: {'A': 1, 'B': 2, 'C': 1} 
- Example Output: {1: ['A', 'C'], 2: ['B']}

```python
# Alternative manual conversion
dict_input = input("Enter a dictionary:").replace('{', '').replace('}', '').replace("'", "").split(',')
dict_converted = {}
for vals in dict_input:
    vals_split = vals.strip().split(':')
    dict_converted[vals_split[0]] = int(vals_split[1])

# Alternative using ast
# import ast
# dict_input = ast.literal_eval(input("Enter a dictionary:"))
# print(dict_input)

values_set = set(dict_converted.values())

final_dict = {}

for val in values_set:
    keys = [k for k, v in dict_converted.items() if v == val]
    final_dict[val] = keys

print(final_dict)
```

15. Create a dictionary where the keys are the vowels found in a user's string and the values are lists of the indices where those vowels appear. 
- Example Input: "banana" 
- Example Output: {'a': [1, 3, 5]}

```python
word_input = input('Enter a word:').lower().strip()
vowels_list = ['a', 'e', 'i', 'o', 'u']
vowels_dict = {}

#Get index for vowels and add to dictionary, create key first if it wasn't created and append indices to its list values.
for i, char in enumerate(word_input):
    if char in vowels_list:
        vowels_dict.setdefault(char, []).append(i)

print(vowels_dict)
```
