1. Write a function that takes a list of strings and a number K. The function should return a new list containing only the strings that have a length greater than K.
- Input: list=["apple", "pie", "banana"], K=3
- Output: ['apple', 'banana']

```python
def filter_length(str_lst, length):
    return [txt for txt in str_lst if len(txt) > length]
```

```python
list_q1 = ['apple', 'pie', 'banana']
k_q1 = 3

filter_length(list_q1, k_q1)
```

2. Write a function named area() that calculates the area of either a "circle" or a
"rectangle". The function should take the shape name as the first argument,
followed by the necessary dimensions (radius for circle; length and width for
rectangle).
- Example Input: area("circle", 5) OR area("rectangle", 4, 6)
- Example Output: Circle Area: 78.5, Rectangle Area: 24

```python
def area(shape, dim1, dim2=None):
    """Calculate the area of a circle or rectangle."""
    shape = shape.strip().lower()
    if dim1 <= 0:
        return "Please enter proper dimensions."
    
    if shape == 'circle':
        return f"Circle Area: {3.14 * dim1**2}"
    elif shape == 'rectangle':
        if dim2:
            return f"Rectangle Area: {dim1 * dim2}"
        else:
            return f"Please add proper width value."
    else:
        return "Please write proper shape name, 'circle' or 'rectangle'"
```

3. Write a program that takes an email address as input and splits it into two parts: the

4. Write a program to find the sum of all numbers in a list. ensure
the program ignores any elements in the list that are not numbers (like strings or
booleans).
- Example Input: [10, "apple", 20, True, 5]
- Example Output: Total Sum: 35

```python
# 1st function: items need to be int or float to be considered in summation (ignoring boolean because it can add 1 for True)
def sum_list1(list_q4):
    print(f"Total Sum:  {sum(x for x in list_q4 if isinstance(x, (int, float)) and not isinstance(x, bool) )}")

# 2nd function: also accepts positive integer values stored as strings,
# but does not handle decimals or negative numbers stored as strings.
def sum_list2(list_q4):
    print(f"Total Sum:  {sum(float(x) for x in list_q4 if str(x).isnumeric() )}")
```

```python
sum_list1([10, "apple", 20, True, 5])
sum_list2([10, "apple", 20, True, 5])
```

```python
sum_list1(["10", "apple", 20, True, 5]) # Doesn't handle the numeric values in string
sum_list2(["10", "apple", 20, True, 5]) # Handles numeric values in string partially
```

5. Write a Python class Restaurant with attributes like menu_items, book_table, and
customer_orders, and methods like add_item_to_menu, book_tables, and
customer_order. (Note: Use dictionaries and lists to store the data.)

```python
class Restaurant:

    def __init__(self):
        # Initialize attributes as private to avoid direct access
        self.__menu_items = {}
        self.__booked_tables = []
        self.__customer_orders = []


    def add_item_to_menu(self, item, price):
        """Add an item and its price to the restaurant menu."""
        self.__menu_items[item] = price


    def book_table(self, customer_name, table_number, guests_count=2):
        """Create and store a table reservation for a customer."""
        reservation = {
            "Customer": customer_name,
            "Table": table_number,
            "Number of Guests": guests_count
        }

        self.__booked_tables.append(reservation)


    def take_order(self, customer_name, items):
        """Create and store an order for a customer."""
        order = {
            "Customer": customer_name,
            "Items": items
        }

        self.__customer_orders.append(order)

    #Attribute getters
    def get_menu_items(self):
        return self.__menu_items

    def get_booked_tables(self):
        return self.__booked_tables

    def get_customer_orders(self):
        return self.__customer_orders

    
```

Perform the following tasks now:
- Now add items to the menu.
- Make table reservations.
- Take customer orders.
- Print the menu.
- Print table reservations.
- Print customer orders.

```python
restaurant = Restaurant()

#Add items to the menu
restaurant.add_item_to_menu("Pasta", 20)
restaurant.add_item_to_menu("Burger", 30)
restaurant.add_item_to_menu("Pizza", 25)
restaurant.add_item_to_menu("Salad", 15)


#Make table reservations
restaurant.book_table("John", 5)
restaurant.book_table("Jane", 1, 4)

#Take customer orders
restaurant.take_order("John", ["Burger", "Pizza"])
restaurant.take_order("Jane", ["Salad"])

#Print the menu
print(restaurant.get_menu_items())

#Print table reservations
print(restaurant.get_booked_tables())

#Print customer orders
print(restaurant.get_customer_orders())
```

### Customer

```python
class Customer:
    """Represent a bike rental customer with a name and age."""
    
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    # Getter for name
    def get_name(self):
        return self.__name

    # Setter for name
    def set_name(self, name):
        self.__name = name

    # Getter for age
    def get_age(self):
        return self.__age

    # Setter for age
    def set_age(self, age):
        self.__age = age
```

### Bike

```python
class Bike:
    """Represent a bike with a name and price."""
    
    def __init__(self, name, price):
        self.__name = name
        self.__price = price

    # Getter for name
    def get_name(self):
        return self.__name

    # Setter for name
    def set_name(self, name):
        self.__name = name

    # Getter for price
    def get_price(self):
        return self.__price

    # Setter for price
    def set_price(self, price):
        self.__price = price
```

### MainBikeRental

```python
class MainBikeRental:
    """Manage available bikes, rentals, bike requests and returns."""

    def __init__(self):
        self.__bikes = []
        self.__rentals = []

    # Getter for bikes
    def get_bikes(self):
        return self.__bikes

    # Setter for bikes
    def set_bikes(self, bikes):
        self.__bikes = bikes

    # Getter for rentals
    def get_rentals(self):
        return self.__rentals

    # Setter for rentals
    def set_rentals(self, rentals):
        self.__rentals = rentals


    def add_bike(self, bike):
        """Add a bike to the available bike stock."""
        self.__bikes.append(bike)


    def request_bike(self, customer, number_of_bikes):
        """Rent bikes to an eligible customer if enough bikes are available."""
        if customer.get_age() <= 6:
            print("Customer must be older than 6 years.")
            return

        if number_of_bikes <= len(self.__bikes):
            rented_bikes = []

            for i in range(number_of_bikes):
                rented_bikes.append(self.__bikes.pop())

            rental = {
                "customer": customer,
                "bikes": rented_bikes
            }

            self.__rentals.append(rental)

            print("Bike rental successful.")

        else:
            print("Wait 10 mins.")


    def return_bike(self, rental_time, num_of_bikes):
        """Return rented bikes to stock and calculate the rental bill."""

        total_bill = rental_time * 40

        for i in range(num_of_bikes):
            returned_bike = Bike("Returned Bike", 0)
            self.__bikes.append(returned_bike)

        print(f"Total bill: {total_bill}")

        return total_bill


    def get_total_cost(self):
        """Return the total price of all bikes currently in the store."""
        
        total = 0

        for bike in self.__bikes:
            total += bike.get_price()

        return total

```

### Classes usage

```python
# Create bike rental shop
shop = MainBikeRental()

# Create bikes
bike1 = Bike("Bike 1", 100)
bike2 = Bike("Bike 2", 200)
bike3 = Bike("Bike 3", 300)

# Add bikes to the store
shop.add_bike(bike1)
shop.add_bike(bike2)
shop.add_bike(bike3)

# Create customer
customer1 = Customer("Ahmed", 20)

# Customer requests 2 bikes
shop.request_bike(customer1, 2)

# Print number of available bikes
print(f"Available bikes: {len(shop.get_bikes())}")

# Return 2 bikes after 3 hours
shop.return_bike(3, 2)

# Print number of available bikes after bikes' return
print(f"Available bikes: {len(shop.get_bikes())}")

# Print total cost of bikes currently in the store
print(f"Total cost: {shop.get_total_cost()}")
```
